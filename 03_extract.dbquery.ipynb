```sql
-- 03_extract
-- Copies all four NorthRiver source tables from northriver_oltp to northriver_staging.
-- Source: northriver_oltp | Target: northriver_staging
-- Run first in the pipeline sequence.
CREATE SCHEMA IF NOT EXISTS northriver_staging
  COMMENT 'Staging copies of NorthRiver OLTP tables, refreshed wholesale on each load';

CREATE TABLE IF NOT EXISTS northriver_staging.stg_customers (
  customer_id BIGINT,
  first_name  VARCHAR(50),
  last_name   VARCHAR(50),
  email       VARCHAR(100),
  region      VARCHAR(50),
  segment     VARCHAR(50),
  updated_at  TIMESTAMP
);

CREATE TABLE IF NOT EXISTS northriver_staging.stg_products (
  product_id   BIGINT,
  product_name VARCHAR(100),
  brand        VARCHAR(50),
  category     VARCHAR(50),
  subcategory  VARCHAR(50),
  unit_cost    DECIMAL(10, 2),
  updated_at   TIMESTAMP
);

CREATE TABLE IF NOT EXISTS northriver_staging.stg_orders (
  order_id    BIGINT,
  customer_id BIGINT,
  order_date  DATE,
  channel     VARCHAR(50),
  updated_at  TIMESTAMP
);

CREATE TABLE IF NOT EXISTS northriver_staging.stg_order_items (
  order_item_id BIGINT,
  order_id      BIGINT,
  product_id    BIGINT,
  quantity      INTEGER,
  unit_price    DECIMAL(10, 2),
  updated_at    TIMESTAMP
);

INSERT OVERWRITE northriver_staging.stg_customers   SELECT * FROM northriver_oltp.stg_customers;
INSERT OVERWRITE northriver_staging.stg_products    SELECT * FROM northriver_oltp.stg_products;
INSERT OVERWRITE northriver_staging.stg_orders      SELECT * FROM northriver_oltp.stg_orders;
INSERT OVERWRITE northriver_staging.stg_order_items SELECT * FROM northriver_oltp.stg_order_items;

SELECT 'stg_customers'   AS tbl, COUNT(*) AS rows FROM northriver_staging.stg_customers   UNION ALL
SELECT 'stg_products',           COUNT(*)         FROM northriver_staging.stg_products    UNION ALL
SELECT 'stg_orders',             COUNT(*)         FROM northriver_staging.stg_orders      UNION ALL
SELECT 'stg_order_items',        COUNT(*)         FROM northriver_staging.stg_order_items
ORDER BY tbl;
```
